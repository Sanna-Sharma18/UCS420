```python
# WAP to check if number is prime or non prime
num = int(input("Enter the number : "))
if num <= 1:
    print("The number", num, "is non prime")
else:
    for i in range(2,num):
        if num % i == 0:
            print("The number", num, "is non prime")
            break
    else:
        print("The number", num, "is prime")
```

```python
# WAP to add all numbers divisible by 7 and 9 from 1 to n and n is given by user
num = int(input("Enter the number : "))
sum = 0
for i in range(1, num+1):
    if i%7 == 0 and i%9 == 0:
        sum = sum + i
print("The sum of all numbers divisible by 7 and 9 from 1 to", num, "is :", sum)
```

```python
# WAP to add all the prime numbers from 1 to n n is given by user
num = int(input("Enter the number : "))
for i in range(2, num + 1):
    prime = True
    for j in range(2, i):
        if i % j == 0:
            prime = False
            break
    if prime:
        sum = sum + i
print("Sum of prime numbers =", sum)
```
