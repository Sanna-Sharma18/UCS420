```python
# Question 2
import numpy as np

arr = np.array([1, 2, 3, 6, 4, 5])

print("Original Array:", arr)
print("Reversed Array:", arr[::-1])

x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

values, counts = np.unique(x, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(x == most_frequent)[0]

print()

print("Most frequent value:", most_frequent)
print("Indices:", indices)

print()

y = np.array([1, 1, 2, 3, 4, 2, 4, 3, 3])

values, counts = np.unique(y, return_counts=True)
most_frequent = values[np.argmax(counts)]
indices = np.where(y == most_frequent)[0]

print("Most frequent value:", most_frequent)
print("Indices:", indices)
```

```python
# Question 4
import numpy as np

Sanna_Sharma = np.linspace(10, 100, 25)

print("Array:")
print(Sanna_Sharma)

print("Dimensions:", Sanna_Sharma.ndim)
print("Shape:", Sanna_Sharma.shape)
print("Total elements:", Sanna_Sharma.size)
print("Data type:", Sanna_Sharma.dtype)
print("Total bytes:", Sanna_Sharma.nbytes)

# Transpose using reshape
transposed = Sanna_Sharma.reshape(1, 25)

print("Transpose using reshape:")
print(transposed)

# Transpose using T
print("Transpose using T:")
print(Sanna_Sharma.T)
```

```python
# Question 5
import numpy as np

# Create 2-D array
ucs420_Sanna_Sharma = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("Original Array:")
print(ucs420_Sanna_Sharma)
print()

# Mean
print("Mean:", np.mean(ucs420_Sanna_Sharma))

# Median
print("Median:", np.median(ucs420_Sanna_Sharma))

# Maximum
print("Maximum:", np.max(ucs420_Sanna_Sharma))

# Minimum
print("Minimum:", np.min(ucs420_Sanna_Sharma))

# Unique elements
print("Unique elements:", np.unique(ucs420_Sanna_Sharma))

# Reshape to 4 rows and 3 columns
ucs420_Sanna_Sharma = ucs420_Sanna_Sharma.reshape(4, 3)

print("\nAfter Reshaping to 4 x 3:")
print(ucs420_Sanna_Sharma)

# Resize to 2 rows and 3 columns
resized_ucs420_Sanna_Sharma = np.resize(ucs420_Sanna_Sharma, (2, 3))

print("\nAfter Resizing to 2 x 3:")
print(resized_ucs420_Sanna_Sharma)
```
