```python
# Question 1
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

temperature = temperature + 2
fahrenheit = (9/5) * temperature + 32

greater = temperature[temperature > 32]
count = np.sum(temperature > 32)

print("Temperatures (after adding 2) :", temperature)
print("Temperature in Fahrenheit :", fahrenheit)
print("Temperature Readings greater than 32 :", greater)
print("Number of readings greater than 32°C :", count)
```

```python
# Question 2
import numpy as np

steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

print("Total steps :", np.sum(steps))
print("Mean steps :", np.mean(steps))

print("Maximum :", np.max(steps))
print("Minimum :", np.min(steps))

print("Total for each day :", np.sum(steps, axis=0))
print("Total for each user :", np.sum(steps, axis=1))

pos=np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position :", pos)
```

```python
# Question 4
import numpy as np

X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

print("Shape of X:", X.shape)
print("Dimensions of X:", X.ndim)
print()

XT = X.T
print("Transpose of X:")
print(XT)
print()

XTX = XT @ X
print("X.T @ X:")
print(XTX)
print()

XTX_inverse = np.linalg.inv(XTX)
print("Inverse of X.T @ X:")
print(XTX_inverse)
print()

beta = XTX_inverse @ XT @ y
print("OLS coefficients:")
print(beta)
print()

print("Coefficients:")
print("Sleep coefficient:", beta[0])
print("Activity coefficient:", beta[1])
print("Stress coefficient:", beta[2])

new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("Predicted assistance score:", predicted_score)
```
