```python
# Question 2
roll_no = "10241794"   # Roll no = 1024170094
L = [int(digit) * 10 for digit in roll_no]
marks = tuple(L[:8])
print("Marks tuple:", marks)
```

```python
# Question 2.1
highest = max(marks)
highest_index = marks.index(highest)

lowest = min(marks)
lowest_count = marks.count(lowest)

print("Highest score:", highest)
print("Index of highest score:", highest_index)
print("Lowest score:", lowest)
print("Number of times lowest score appears:", lowest_count)
```

```python
# Question 2.2
# Tuples are immutable so their elements cannot be changed or reversed in place 
# and a new reversed list must be created.

reversed_scores = [x for x in reversed(marks)]
print("Reversed scores:", reversed_scores)
```

```python
# Question 2.3
m = int(input("Enter a score: "))
if m in marks:
    print("First occurrence index :", marks.index(m))
else:
    print("Score not present in the tuple")
```

```python
# Question 2.4
marks[0] = 100
# A tuple is immutable, so its elements cannot be changed directly, unlike a list.
```

```python
# Question 2.5
first_score, second_score, *remaining_scores = marks

print("First score :", first_score)
print("Second score :", second_score)
print("Remaining scores :", remaining_scores)
```

```python
# Question 3.1
import random
roll_number = 1024170094
random.seed(roll_number)

numbers = [random.randint(100, 900) for _ in range(100)]
print("List of 100 random numbers :")
print(numbers)
```

```python
# Question 3.2
odd_numbers = [num for num in numbers if num % 2 != 0]

print("Odd numbers:")
print(odd_numbers)
print("Count of odd numbers:", len(odd_numbers))
```

```python
# Question 3.3
even_numbers = [num for num in numbers if num % 2 == 0]

print("Even numbers:")
print(even_numbers)
print("Count of even numbers:", len(even_numbers))
```

```python
# Question 3.4
def is_prime(num):
    if num <= 1:
        return False
    else:
        for i in range(2,num):
            if num % i == 0:
                return False
        return True

prime_numbers = [num for num in numbers if is_prime(num)]

print("Prime numbers :")
print(prime_numbers)

print("Count of prime numbers:", len(prime_numbers))

prime_list = [num for num in numbers if is_prime(num)]
print("Prime list using list comprehension:")
print(prime_list)
```

```python
# Question 3.5
frequency = {}

for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1

most_frequent = max(frequency, key=frequency.get)

print("Most frequently occurring number:", most_frequent)
print("Number of times it occurs:", frequency[most_frequent])
```

```python
# Question 4
roll_no = "10241794"   # Roll no = 1024170094
set_A = {int(digit) * 7 for digit in roll_no}
set_B = {int(digit) * 9 for digit in roll_no}

print("Set A :", set_A)
print("Set B :", set_B)
```

```python
# Question 4.6
union_AB = set_A.union(set_B)

print("Union of A and B :", union_AB)
```

```python
# Question 4.7
intersection_AB = set_A.intersection(set_B)

print("intersection of A and B :", intersection_AB)
```

```python
# Question 4.8
difference_AB = set_A.difference(set_B)
difference_BA = set_B.difference(set_A)

print("A - B :", difference_AB)

print("B - A :", difference_BA)
```

```python
# Question 4.9
symmetric_AB = set_A.symmetric_difference(set_B)

print("Symmetric difference of A and B :", symmetric_AB)

# difference() gives elements present in one set but not in the other,
# while symmetric_difference() gives elements unique to both sets together
```

```python
# Question 4.10
print("Is A a subset of B? :", set_A.issubset(set_B))

print("Is B a superset of A? :", set_B.issuperset(set_A))
```

```python
# Question 4.11
x = int(input("Enter a value X to remove from Set A: "))

set_A.discard(x)

print("Set A after discard() :", set_A)

# discard() is safer than remove() because it does not raise an error
# if the specified value is not present in the set 
# whereas remove() raises an error when the element is not present
```

```python
# Question 5.1
my_dict = { 
"name" : "Sanna Sharma", 
"roll_no" : "1024170094", 
"branch" : "COPC", 
"age": 20, 
"city": "Patiala" 
} 

print("Original dictionary :")
print(my_dict)
```

```python
# Question 5.1
city_value = my_dict.pop("city")
my_dict["location"] = city_value

print("After renaming city to location:")
print(my_dict)
```

```python
# Question 5.2
my_dict["cgpa"] = 8.85

print("After adding CGPA :")
print(my_dict)
```

```python
# Question 5.3
my_dict["age"] = my_dict["age"] + 1

print("After increasing age by 1 :")
print(my_dict)
```

```python
# Question 5.4
# Using pop()
dict_pop = my_dict.copy()
remove_branch = dict_pop.pop("branch")

print("Using pop() :")
print(dict_pop)
print("Value returned by pop() :", remove_branch)

# Using del
dict_del = my_dict.copy()
del dict_del["branch"]

print("Using del :")
print(dict_del)

# pop() removes the key and returns its value,
# while del removes the key but does not return its value.
```

```python
# Question 5.5
print("Key value pairs of dictionary :")

for key, value in my_dict.items():
    print(key, "-->", value)
```

```python
# Question 5.6
print("Checking for email :")

if "email" in my_dict:
    print("Email:", my_dict["email"])
else:
    print("Email key is not present in the dictionary")
```

```python
# Question 5.7

# When both dictionaries have the same key, the value from the second dictionary (friend_dict) wins
friend_dict = {
    "name": "Varun",
    "roll_no": "1024171111",
    "branch": "CSE",
    "age": 22,
    "city": "Allahabad"
}

merged_dict = {**my_dict, **friend_dict}

print("Friend dictionary:")
print(friend_dict)

print("\nMerged dictionary:")
print(merged_dict)
```
