```python
# Question 1

import pandas as pd
roll_no = "94"

last_two = [int(d) for d in roll_no[-2:]]

categories = ["billing", "account", "general"]

faq1 = {
    "category": categories[last_two[0] % 3],
    "question": "How can I check the status of my order?",
    "answer": "You can check your order status from the Orders section of your account.",
    "keywords": ["order", "status", "tracking"]
}

faq2 = {
    "category": categories[last_two[1] % 3],
    "question": "How can I download my payment receipt?",
    "answer": "You can download your payment receipt from the Billing section of your account.",
    "keywords": ["payment", "receipt", "billing"]
}

fixed_faqs = [
    {"question": "what is the annual fee", "answer": "The annual fee is Rs 500.",  "keywords": "fee cost price charge", "category": "billing"}, 
    {"question": "how to reset password", "answer": "Go to Settings > Reset Password.",  "keywords": "password reset login", "category": "account"}, 
    {"question": "what are your working hours", "answer": "We are open 9 AM to 5 PM.",  "keywords": "hours timing open time", "category": "general"}, 
    {"question": "how can i pay the fee", "answer": "You can pay via UPI, card, or net  banking.", 
    "keywords": "pay payment upi fee", "category": "billing"},
]

faq_data = fixed_faqs + [faq1, faq2]

df = pd.DataFrame(faq_data)

print(df)
```

```python
# Question 2 

import re

def score_hypothesis(query, df):
    query_words = set(re.findall(r'\w+', query.lower()))
    results = []

    for _, row in df.iterrows():

        keywords = set(row["keywords"])
        matched_words = query_words.intersection(keywords)


        if len(keywords) > 0:
            confidence = len(matched_words) / len(keywords)
        else:
            confidence = 0

        if confidence > 0:
            results.append({
                "category": row["category"],
                "question": row["question"],
                "answer": row["answer"],
                "confidence": round(confidence, 2)
            })

    results.sort(key=lambda x: x["confidence"], reverse=True)

    return results

query = "How can I check my payment receipt?"

results = score_hypothesis(query, df)

for result in results:
    print(result)
```

```python
# Question 6

def score_query(query, df):
    query_words = query.lower().split()
    results = []

    for index, row in df.iterrows():
        keywords = row["keywords"].lower().split()

        score = sum(word in keywords for word in query_words)

        if score > 0:
            results.append((index, score))

    if not results:
        return []

    max_score = max(score for index, score in results)

    best_results = [
        (index, score)
        for index, score in results
        if score == max_score
    ]

    return best_results
```
